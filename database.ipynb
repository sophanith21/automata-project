### read data from json file and insert data into the database

```python
import json

with open('designFA.json', 'r') as file:
    data = json.load(file)

name = data['fa_name']
faType = data['fa_type']
numOfStates = data['numOfStates']
numOfSymbols = data['numOfSymbols']
states = data['states']
statesString = json.dumps(states);  # this is used to convert from python format to json format so that we can store them into the database
symbols = data['symbols']
symbolsString = json.dumps(symbols)
transition = data['transition']
transitionString = json.dumps(transition)
print(name);
print(faType);
print(numOfStates);
print(numOfSymbols);
print(states);
print(states);
print(transition );

cursor = connection.cursor()

insertQuery = """ 
    INSERT INTO FA(name, numberOfState, numberOfSymbol, state, symbol, faType, transition)
    VALUES(%s, %s, %s, %s, %s, %s, %s)
"""

insertData =(name, numOfStates, numOfSymbols, statesString, symbolsString, faType, transitionString)
cursor.execute(insertQuery, insertData)
connection.commit()
print("save to database successfully")


```
